## QUESTION 2: Neha’s Salary Shock

```python
n = int(input())
records = []
for _ in range(n):
    records.append(input().split()) # id, hour, amount

affected = set()
for i in range(n):
    id1, h1, a1 = records[i][0], int(records[i][1]), float(records[i][2])
    if a1 > 0:
        for j in range(n):
            id2, h2, a2 = records[j][0], int(records[j][1]), float(records[j][2])
            if id1 == id2 and a2 == -a1 and 0 < (h2 - h1) <= 24:
                affected.add(id1)
print(*(sorted(list(affected))))
```

## QUESTION 3: The Curious Case of the Copy Paste

```python
import re

def get_words(text):
    return re.findall(r'\w+', text.lower())

n = int(input())
resumes = [get_words(input()) for _ in range(n)]
found = False

for i in range(n):
    for j in range(i + 1, n):
        words1, words2 = set(resumes[i]), set(resumes[j])
        common = words1.intersection(words2)
        match_rate = len(common) / max(len(words1), len(words2))
        if match_rate >= 0.8:
            print(f"{i}{j}")
            found = True
if not found: print("No duplicates")
```

## QUESTION 14: The Exam Seating Chaos

```python
n = int(input())
rolls = list(map(int, input().split()))
odds = [x for x in rolls if x % 2 != 0]
evens = [x for x in rolls if x % 2 == 0]
res = []
for o, e in zip(odds, evens):
    res.extend([o, e])
# Add remaining if lengths differ
res.extend(odds[len(evens):] or evens[len(odds):])
print(*res)
```

## QUESTION 16: Festival Crowd Panic

```python
n = int(input())
events = []
for _ in range(n):
    en, ex = map(int, input().split())
    events.append((en, 1))
    events.append((ex, -1))
events.sort()
max_p, current_p = 0, 0
for _, change in events:
    current_p += change
    max_p = max(max_p, current_p)
print(max_p)
```

## 27. The Cyber Attack Window

```python
n, k, s = map(int, input().split())
packets = list(map(int, input().split()))
found = "NO"
curr_sum = sum(packets[:k])
if curr_sum > s: found = "YES"
else:
    for i in range(k, n):
        curr_sum += packets[i] - packets[i-k]
        if curr_sum > s:
            found = "YES"
            break
print(found)
```
